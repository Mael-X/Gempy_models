```python
import os

os.environ["DEFAULT_BACKEND"] = "PYTORCH"

import gempy as gp # type: ignore
import gempy_viewer as gpv # type: ignore
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd # type: ignore
import pyvista as pv
#Ignore car libraries bien importées !

# print("gempy version: ", gp.__version__)
# print("gempy_viewer version: ", gpv.__version__)
```

```python
tabular_model = gp.create_geomodel(
    project_name="TabularModel",
    extent=[0, 1000, 0, 1000, 0, 1000],
    refinement = 4,
    # resolution = [50, 50, 50],
    structural_frame = gp.data.StructuralFrame.initialize_default_structure()
) 

#Rename, recolor and append points and surface orientation to the default surface1

tabular_model.structural_frame.structural_elements[0].name = "Tithonic"  # Renaming 'surface 1' to 'Tithonic'.
tabular_model.structural_frame.structural_elements[0].color = '#00FFFF'  # Bright Teal

position = gp.add_surface_points(
    geo_model = tabular_model,
    x = np.array([100,900]),  # X position
    y = np.array([0,0]),      # Y position
    z = np.array([200,200]),  # Z position
    elements_names = ["Tithonic", "Tithonic"]

)

orientation = gp.add_orientations(
    geo_model = tabular_model,
    x = np.array([100]),  # X position
    y = np.array([0]),    # Y position
    z = np.array([200]),  # Z position
    elements_names = ["Tithonic"],
    pole_vector=np.array([0, 0, 1]) # Vertical pole vector for the horizontal plane

                                )

#Defining the other layers

element2 = gp.data.StructuralElement(name = "Berriasian",
                                    color = '#006400',  #Dark Green
                                    surface_points=gp.data.SurfacePointsTable.from_arrays
                                    (
                                        np.array([100, 900]),  # X coordinates
                                        np.array([0, 0]),  # Y coordinates
                                        np.array([400, 400]),   # Z coordinates
                                        names = ["Berriasian", "Berriasian"],
                                    ),
                                    orientations = gp.data.OrientationsTable.initialize_empty()
                                )

tabular_model.structural_frame.structural_groups[0].append_element(element2)

element3 = gp.data.StructuralElement(name = "Valanginian",
                                    color = '#008000',  # Slight Light Green
                                    surface_points=gp.data.SurfacePointsTable.from_arrays
                                    (
                                        np.array([100, 900]),  # X coordinates
                                        np.array([0, 0]),  # Y coordinates
                                        np.array([600, 600]),   # Z coordinates
                                        names = ["Valanginian", "Valanginian"],
                                    ),
                                    orientations = gp.data.OrientationsTable.initialize_empty()
                                )

tabular_model.structural_frame.structural_groups[0].append_element(element3)

element4 = gp.data.StructuralElement(name = "Kimmeridgian",
                                    color = '#00FF00',  # Bright Green
                                    surface_points=gp.data.SurfacePointsTable.from_arrays
                                    (
                                        np.array([100, 900]),  # X coordinates
                                        np.array([0, 0]),  # Y coordinates
                                        np.array([750, 750]),  # Z coordinates
                                        names = ["Kimmeridgian", "Kimmeridgian"],
                                    ),
                                    orientations = gp.data.OrientationsTable.initialize_empty()
                                )

tabular_model.structural_frame.structural_groups[0].append_element(element4)

#One more orientation added to be sure the model cumputes flat surfaces

orientations=gp.add_orientations(
    geo_model = tabular_model,
    x = np.array([100]),  # X position
    y = np.array([0]),  # Y position
    z = np.array([750]),  # Z position
    elements_names = ["Kimmeridgian"],
    pole_vector=np.array([0, 0, 1]) # Vertical pole vector for the horizontal plane
                                     )


tabular_model.structural_frame.structural_groups[0].name = "Stratigraphic. Units"


# tabular_model.structural_frame.structural_elements[0]

tabular_model.structural_frame #uncomment to see the structural frame
```

```python
#plot the model in 2D in XZ plane
fig = gpv.plot_3d(
    tabular_model,
    show_data_points=True,
    show_orientations=True,
    show_title=True,
    show_legend=True
)
```
